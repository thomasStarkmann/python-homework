```python
import numpy as np
data = np.loadtxt("C:/Users/thoma/Desktop/swc-python/data/inflammation-01.csv", delimiter = ",")
```

```python
import matplotlib.pyplot as plt
image = plt.imshow(data)
plt.show()
```

```python
ave_inflammation = np.mean(data, axis=0)
ave_plot = plt.plot(ave_inflammation)
plt.show()
```

```python
max_plot = plt.plot(np.max(data, axis=0))
plt.show()
```

```python
fig = plt.figure(figsize=(10.0, 3.0))

axes1 = fig.add_subplot(1, 3, 1)
axes2 = fig.add_subplot(1, 3, 2)
axes3 = fig.add_subplot(1, 3, 3)

axes1.set_ylabel("average")
axes1.plot(np.mean(data, axis=0), drawstyle="steps-mid")
axes1.set_ylim(0, 21)

axes2.set_ylabel("max")
axes2.plot(np.max(data, axis=0), drawstyle="steps-mid")
axes2.set_ylim(0, 21)

axes3.set_ylabel("min")
axes3.plot(np.min(data, axis=0), drawstyle="steps-mid")
axes3.set_ylim(0, 21)

fig.tight_layout()

plt.savefig("inflammation.png")
plt.show()
```

```python
std_plot = plt.plot(np.std(data, axis=0), drawstyle="steps-mid")
plt.show()
```
