Problem
Given: Two positive integers a and b (a<b<10000).

Return: The sum of all odd integers from a through b, inclusively.

Sample Dataset
100 200

Sample Output
7500

```python
a = 100
b = 200

total = 0
for number in range(a, b):
    if number % 2 == 1:
        total = total + number
print(total)

```
